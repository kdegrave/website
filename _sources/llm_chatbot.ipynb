# An LLM-Powered Chatbot

In this post, we'll walk through the steps to build an LLM-powered chat interface using Python. This simple application leverages tools like OpenAI's API, Gradio, and environment variables. I'll guide you through the code step-by-step, explaining the purpose of each part.

## Import Required Libraries

The first step in any Python project is to import the necessary libraries.

```python
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
import os
```

* `dotenv`: Used for managing environment variables securely.
* `openai`: The library for interacting with OpenAI’s API.
* `gradio`: A Python library for building user interfaces quickly.
* `os`: Allows us to access environment variables in our operating system.

We use the `dotenv` library to load our environment variables securely. The API key for OpenAI is stored in an `.env` file and accessed through Python's `os.getenv` method. This ensures sensitive information stays safe.

```python
load_dotenv()

openai_api_key = os.getenv('OPENAI_API_KEY')
```

Next, we initialize the OpenAI API with the desired model. In this example, we’re using `gpt-4o-mini`, a smaller and faster variant of GPT-4.

```python
openai = OpenAI()

model = 'gpt-4o-mini'
```

The system message sets the tone for the AI's responses. It defines the role the AI will play in the chat. Here, we make the assistant a "helpful" one.

```python
system_message = 'You are a helpful assistant.'
```

Here’s where the magic happens! The `chat` function processes user input, sends it to OpenAI's API, and streams back the response.

```python
def chat(message, history):

    messages = [{'role': 'system', 'content': system_message}] + history + [{'role': 'user', 'content': message}]

    stream = openai.chat.completions.create(
        model=model,
        messages=messages,
        stream=True
    )

    response = ''
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

* `message`: The user’s input to the chat interface.
* `history`: Keeps track of previous exchanges to maintain context.
* **Streaming Responses**: Instead of waiting for the entire response, the stream parameter allows us to send partial outputs for a smoother experience.

We use Gradio’s `ChatInterface` to create a user-friendly chat interface. A simple line of code is all it takes to launch the app locally.

```python
gr.ChatInterface(fn=chat, type='messages').launch()
```
